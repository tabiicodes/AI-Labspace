```python
!pip install -U google-genai

from google import genai
from google.colab import userdata

client = genai.Client(
    api_key=userdata.get("STUDIO_API_KEY")
)


MODEL = "models/gemini-3.5-flash"


def travel_agent(user_input):

    prompt = f"""
You are a professional travel agent for Pakistan tourism.

Provide a well-structured travel plan with clear headings:

1. Destination Overview
2. Budget Travel Options (Bus, Train, Flight with approx cost in PKR)
3. Places to Visit (specific locations)
4. Food Suggestions (local dishes)
5. Money-Saving Tips

Make response detailed but easy for students.

User request: {user_input}

Travel Plan:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text


print("Gemini Travel Agent Started (type 'exit' to stop)\n")

while True:
    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye")
        break

    try:
        reply = travel_agent(user)
        print("\nAgent:\n", reply, "\n")

    except Exception as e:
        print("Error:", e)
```

```python
!pip install -U google-genai

from google import genai
from google.colab import userdata

client = genai.Client(
    api_key=userdata.get("STUDIO_API_KEY")
)

MODEL = "models/gemini-3.5-flash"


def smart_finder(user_input):

    prompt = f"""
You are Smart Finder AI, an intelligent lost-and-found matching agent.

Your task is to analyze a user's lost item description and determine
whether it could match a found item.

Consider these important matching features:

1. Item type
2. Brand
3. Color
4. Model or identifying features
5. Location
6. Date or time
7. Unique characteristics

Provide your response with these clear headings:

1. Lost Item Details
2. Found Item Details
3. Matching Features
4. Match Result
5. Confidence Level
6. Explanation

The Match Result should be one of:
- Strong Match
- Possible Match
- No Match

Do not claim that an item definitely belongs to the user.
Give a reasonable match based only on the information provided.

User's lost-and-found information:
{user_input}

Smart Finder Analysis:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text


print("Smart Finder AI Started (type 'exit' to stop)\n")

while True:
    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye")
        break

    try:
        reply = smart_finder(user)
        print("\nSmart Finder AI:\n", reply, "\n")

    except Exception as e:
        print("Error:", e)
```
